```python
import numpy as np
import matplotlib.pyplot as plt

# Function to generate Unit Impulse Signal
def unit_impulse(length, position):
    signal = np.zeros(length)
    signal[position] = 1
    return signal

# Parameters
start = -10
stop = 10
step = 1

# X-axis values
x = np.arange(start, stop + step, step)

# Generate signal
impulse_signal = unit_impulse(len(x), abs(start) // step)

# Plot
plt.figure(figsize=(8,4))
plt.stem(x, impulse_signal)
plt.title("Unit Impulse Signal")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Function to generate Impulse Train
def impulse_train(signal_length, period):
    signal = np.zeros(signal_length)

    for i in range(signal_length):
        if i % period == 0:
            signal[i] = 1

    return signal

# Parameters
signal_length = 100
period = 10

# Generate Impulse Train
x = np.arange(signal_length)
y = impulse_train(signal_length, period)

# Plot
plt.figure(figsize=(10,4))
plt.stem(x, y)
plt.title("Impulse Train Signal")
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Continuous Unit Step Signal
t = np.linspace(-5, 5, 1000)
continuous_step = np.where(t >= 0, 1, 0)

# Discrete Unit Step Signal
n = np.arange(0, 20)
discrete_step = np.where(n >= 0, 1, 0)

# Plot
plt.figure(figsize=(10,6))

# Continuous
plt.subplot(2,1,1)
plt.plot(t, continuous_step, linewidth=2)
plt.title("Continuous Unit Step Signal")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)

# Discrete
plt.subplot(2,1,2)
plt.stem(n, discrete_step)
plt.title("Discrete Unit Step Signal")
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Continuous Ramp Signal
t = np.linspace(-5, 5, 1000)
continuous_ramp = np.where(t >= 0, t, 0)

# Discrete Ramp Signal
n = np.arange(0, 20)
discrete_ramp = n

# Plot
plt.figure(figsize=(10,6))

# Continuous Ramp
plt.subplot(2,1,1)
plt.plot(t, continuous_ramp, linewidth=2)
plt.title("Continuous Ramp Signal")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)

# Discrete Ramp
plt.subplot(2,1,2)
plt.stem(n, discrete_ramp)
plt.title("Discrete Ramp Signal")
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.grid(True)

plt.tight_layout()
plt.show()
```
