```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Load the image in grayscale
image = cv2.imread("exp_11_image.jpg", 0)

# Check if image is loaded
if image is None:
    print("Error: Image not found!")
    exit()

# 2. Define the structuring element
kernel = np.ones((5, 5), np.uint8)

# 3. Perform Erosion
erosion = cv2.erode(image, kernel, iterations=1)

# 4. Perform Dilation
dilation = cv2.dilate(image, kernel, iterations=1)

# 5. Perform Opening
# Erosion followed by Dilation
opening = cv2.morphologyEx(
    image,
    cv2.MORPH_OPEN,
    kernel
)

# 6. Perform Closing
# Dilation followed by Erosion
closing = cv2.morphologyEx(
    image,
    cv2.MORPH_CLOSE,
    kernel
)

# Display all results in proper format
plt.figure(figsize=(12, 7))

# Original Image
plt.subplot(2, 3, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image", fontsize=12)
plt.axis("off")

# Erosion
plt.subplot(2, 3, 2)
plt.imshow(erosion, cmap="gray")
plt.title("Erosion", fontsize=12)
plt.axis("off")

# Dilation
plt.subplot(2, 3, 3)
plt.imshow(dilation, cmap="gray")
plt.title("Dilation", fontsize=12)
plt.axis("off")

# Opening
plt.subplot(2, 3, 4)
plt.imshow(opening, cmap="gray")
plt.title("Opening", fontsize=12)
plt.axis("off")

# Closing
plt.subplot(2, 3, 5)
plt.imshow(closing, cmap="gray")
plt.title("Closing", fontsize=12)
plt.axis("off")

# Remove the empty sixth subplot
plt.subplot(2, 3, 6)
plt.axis("off")

plt.suptitle(
    "Experiment-11: Morphological Image Processing",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout(rect=[0, 0, 1, 0.94])
plt.show()
```
