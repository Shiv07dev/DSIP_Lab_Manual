# Experiment 6 — FIR Filter Design Using Windowing Method

**AIM:** Design FIR filters using the windowing method.

**Given:** Fs = 10 kHz, N = 31 taps, cutoff = 3 kHz for LPF/HPF, and 3–6 kHz for BPF/BSF.  
**Window:** Hann.

The supplied experiment material gives the windowing relation `h(n) = hd(n)w(n)` and the SciPy `firwin()` approach with a window. fileciteturn0file0L34-L40 fileciteturn0file0L47-L58

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import firwin, freqz

print("Libraries imported successfully.")
```

```python
# Common parameters
Fs = 10000       # Sampling frequency (Hz)
N = 31           # Number of FIR taps
Fc = 3000        # Cutoff frequency (Hz)
Fc1 = 3000       # Lower cutoff (Hz)
Fc2 = 6000       # Upper cutoff (Hz)
WINDOW = 'hann'

print(f"Fs = {Fs} Hz")
print(f"N = {N} taps")
print(f"Fc = {Fc} Hz")
print(f"Band cutoffs = {Fc1} Hz, {Fc2} Hz")
print(f"Window = {WINDOW}")
```

## 1. FIR Low-Pass Filter — Fc = 3 kHz

```python
# FIR Low-Pass Filter
b_lpf = firwin(
    numtaps=N,
    cutoff=Fc,
    window=WINDOW,
    pass_zero='lowpass',
    fs=Fs
)

f_lpf, H_lpf = freqz(b_lpf, worN=2048, fs=Fs)
mag_lpf_db = 20*np.log10(np.maximum(np.abs(H_lpf), 1e-10))

print("Low-pass FIR coefficients:")
print(np.round(b_lpf, 6))

plt.figure(figsize=(10,5))
plt.plot(f_lpf, mag_lpf_db)
plt.axvline(Fc, linestyle='--', label='Fc = 3 kHz')
plt.title('FIR Low-Pass Filter Frequency Response')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Magnitude (dB)')
plt.xlim(0, Fs/2)
plt.grid(True)
plt.legend()
plt.show()
```

### LPF Output
**Paste/screenshot the graph above here.**

**Observation:** Frequencies below approximately 3 kHz are passed and frequencies above the cutoff are attenuated.

## 2. FIR High-Pass Filter — Fc = 3 kHz

```python
# FIR High-Pass Filter
b_hpf = firwin(
    numtaps=N,
    cutoff=Fc,
    window=WINDOW,
    pass_zero='highpass',
    fs=Fs
)

f_hpf, H_hpf = freqz(b_hpf, worN=2048, fs=Fs)
mag_hpf_db = 20*np.log10(np.maximum(np.abs(H_hpf), 1e-10))

print("High-pass FIR coefficients:")
print(np.round(b_hpf, 6))

plt.figure(figsize=(10,5))
plt.plot(f_hpf, mag_hpf_db)
plt.axvline(Fc, linestyle='--', label='Fc = 3 kHz')
plt.title('FIR High-Pass Filter Frequency Response')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Magnitude (dB)')
plt.xlim(0, Fs/2)
plt.grid(True)
plt.legend()
plt.show()
```

### HPF Output
**Paste/screenshot the graph above here.**

**Observation:** Frequencies below approximately 3 kHz are attenuated and frequencies above the cutoff are passed.

## 3. FIR Band-Pass Filter — Fc1 = 3 kHz, Fc2 = 6 kHz

```python
# FIR Band-Pass Filter
b_bpf = firwin(
    numtaps=N,
    cutoff=[Fc1, Fc2],
    window=WINDOW,
    pass_zero='bandpass',
    fs=Fs
)

f_bpf, H_bpf = freqz(b_bpf, worN=2048, fs=Fs)
mag_bpf_db = 20*np.log10(np.maximum(np.abs(H_bpf), 1e-10))

print("Band-pass FIR coefficients:")
print(np.round(b_bpf, 6))

plt.figure(figsize=(10,5))
plt.plot(f_bpf, mag_bpf_db)
plt.axvline(Fc1, linestyle='--', label='Fc1 = 3 kHz')
plt.axvline(Fc2, linestyle='--', label='Fc2 = 6 kHz')
plt.title('FIR Band-Pass Filter Frequency Response')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Magnitude (dB)')
plt.xlim(0, Fs/2)
plt.grid(True)
plt.legend()
plt.show()
```

### BPF Output
**Paste/screenshot the graph above here.**

**Observation:** The frequency band from approximately 3 kHz to 6 kHz is passed, while frequencies outside this band are attenuated.

## 4. FIR Band-Stop Filter — Fc1 = 3 kHz, Fc2 = 6 kHz

```python
# FIR Band-Stop Filter
b_bsf = firwin(
    numtaps=N,
    cutoff=[Fc1, Fc2],
    window=WINDOW,
    pass_zero='bandstop',
    fs=Fs
)

f_bsf, H_bsf = freqz(b_bsf, worN=2048, fs=Fs)
mag_bsf_db = 20*np.log10(np.maximum(np.abs(H_bsf), 1e-10))

print("Band-stop FIR coefficients:")
print(np.round(b_bsf, 6))

plt.figure(figsize=(10,5))
plt.plot(f_bsf, mag_bsf_db)
plt.axvline(Fc1, linestyle='--', label='Fc1 = 3 kHz')
plt.axvline(Fc2, linestyle='--', label='Fc2 = 6 kHz')
plt.title('FIR Band-Stop Filter Frequency Response')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Magnitude (dB)')
plt.xlim(0, Fs/2)
plt.grid(True)
plt.legend()
plt.show()
```

### BSF Output

**Observation:** The frequency band from approximately 3 kHz to 6 kHz is attenuated, while frequencies outside this band are passed.

## Combined Frequency Response

```python
plt.figure(figsize=(12,7))
plt.plot(f_lpf, mag_lpf_db, label='Low-Pass')
plt.plot(f_hpf, mag_hpf_db, label='High-Pass')
plt.plot(f_bpf, mag_bpf_db, label='Band-Pass')
plt.plot(f_bsf, mag_bsf_db, label='Band-Stop')
plt.axvline(3000, linestyle='--', alpha=0.6, label='3 kHz')
plt.axvline(6000, linestyle='--', alpha=0.6, label='6 kHz')
plt.title('Comparison of FIR Filter Frequency Responses')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Magnitude (dB)')
plt.xlim(0, Fs/2)
plt.ylim(-100, 5)
plt.grid(True)
plt.legend()
plt.show()
```

## Final Result

All four FIR filters were successfully designed using the **windowing method** with **N = 31**, **Fs = 10 kHz**, and a **Hann window**.

- **LPF:** passes frequencies below 3 kHz.
- **HPF:** passes frequencies above 3 kHz.
- **BPF:** passes frequencies from 3 kHz to 6 kHz.
- **BSF:** attenuates frequencies from 3 kHz to 6 kHz.
