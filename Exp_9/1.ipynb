```python
import cv2
import numpy as np
import requests
import matplotlib.pyplot as plt

def cv2_imshow(image):
    if image is None:
        raise ValueError("Image not loaded. Please check the file path.")
    if image.ndim == 2:
        plt.imshow(image, cmap="gray")
    else:
        image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        plt.imshow(image_rgb)
    plt.axis("off")
    plt.show()

# Download image automatically from Unsplash
url = "https://images.unsplash.com/photo-1500530855697-b586d89ba3ee?w=800"
response = requests.get(url)

with open("input_image.jpg", "wb") as f:
    f.write(response.content)

# Load the image
image = cv2.imread("input_image.jpg")

# Display original image
print("Original Image:")
cv2_imshow(image)

# -----------------------------------
# Gaussian Smoothing
# -----------------------------------

kernel_size = (5, 5)
sigma = 1.5

gaussian_kernel = cv2.getGaussianKernel(kernel_size[0], sigma)
gaussian_kernel = np.outer(gaussian_kernel, gaussian_kernel)

# Apply Gaussian smoothing
smoothed_image = cv2.filter2D(image, -1, gaussian_kernel)

print("Gaussian Smoothed Image:")
cv2_imshow(smoothed_image)

# -----------------------------------
# Sharpening
# -----------------------------------

sharpening_kernel = np.array([
    [-1, -1, -1],
    [-1,  9, -1],
    [-1, -1, -1]
])

# Apply sharpening filter
sharpened_image = cv2.filter2D(
    image,
    -1,
    sharpening_kernel
)

print("Sharpened Image:")
cv2_imshow(sharpened_image)
```

```python
import cv2
import numpy as np
import requests


# Download image automatically from Unsplash
url = "https://images.unsplash.com/photo-1500530855697-b586d89ba3ee?w=800"
response = requests.get(url)

with open("input_image.jpg", "wb") as f:
    f.write(response.content)

# Load the image
image = cv2.imread("input_image.jpg")

# Display original image
print("Original Image:")
cv2_imshow(image)

# Define size of averaging filter
kernel_size = (5, 5)

# Create averaging filter kernel
kernel = np.ones(
    kernel_size,
    dtype=np.float32
) / (kernel_size[0] * kernel_size[1])

# Apply averaging filter
smoothed_image = cv2.filter2D(
    image,
    -1,
    kernel
)

# Display smoothed image
print("Image After Averaging Filter:")
cv2_imshow(smoothed_image)
```

```python
import cv2
import numpy as np
import requests


# Download image automatically from Unsplash
url = "https://images.unsplash.com/photo-1500530855697-b586d89ba3ee?w=800"
response = requests.get(url)

with open("input_image.jpg", "wb") as f:
    f.write(response.content)

# Load the image
image = cv2.imread("input_image.jpg")

# Display original image
print("Original Image:")
cv2_imshow(image)

# Apply Gaussian smoothing to reduce noise
blurred_image = cv2.GaussianBlur(
    image,
    (5, 5),
    0
)

print("Blurred Image:")
cv2_imshow(blurred_image)

# Define spatial high-pass / Laplacian sharpening kernel
laplacian_kernel = np.array([
    [ 0, -1,  0],
    [-1,  5, -1],
    [ 0, -1,  0]
], dtype=np.float32)

# Apply sharpening filter
sharpened_image = cv2.filter2D(
    blurred_image,
    -1,
    laplacian_kernel
)

# Display sharpened image
print("Sharpened Image:")
cv2_imshow(sharpened_image)
```
