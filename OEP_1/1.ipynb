```python
# Import NumPy library
import numpy as np

# Import Matplotlib library
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Generate Sample Points
# ------------------------------------------------------------
# Generate points from -2 to 2
n = np.linspace(-2, 2, 500)

# ------------------------------------------------------------
# Define Piecewise Signal
# ------------------------------------------------------------
x = np.piecewise(
    n,
    [
        (-2 <= n) & (n < -1),   # Constant part
        (-1 <= n) & (n < 1),    # Linear ramp
        (1 <= n) & (n < 2),     # Constant part
        (n >= 2)                # Zero after n=2
    ],
    [
        -2,
        lambda n: 2 * n,
        2,
        0
    ]
)

# ------------------------------------------------------------
# Create Figure
# ------------------------------------------------------------
plt.figure(figsize=(8,5))

# ------------------------------------------------------------
# Plot Signal
# ------------------------------------------------------------
plt.plot(n, x, color='blue', linewidth=2)

# Draw vertical lines exactly like the given figure
plt.vlines(-2, 0, -2, colors='blue', linewidth=2)
plt.vlines(2, 2, 0, colors='blue', linewidth=2)

# ------------------------------------------------------------
# Draw X and Y Axes
# ------------------------------------------------------------
plt.axhline(0, color='black')
plt.axvline(0, color='black')

# ------------------------------------------------------------
# Labels and Title
# ------------------------------------------------------------
plt.title("Signal 3 : Piecewise Linear Signal")
plt.xlabel("Sample Index (n)")
plt.ylabel("Amplitude x[n]")

# Student Name
plt.text(1.2, 2.15, "Shivkumar Paun", fontsize=10)

# ------------------------------------------------------------
# Axis Limits
# ------------------------------------------------------------
plt.xlim(-2.3, 2.3)
plt.ylim(-2.5, 2.5)

# ------------------------------------------------------------
# Grid
# ------------------------------------------------------------
plt.grid(True, linestyle='--', alpha=0.7)

# ------------------------------------------------------------
# Display Graph
# ------------------------------------------------------------
plt.show()
```

```python
# Import NumPy library
import numpy as np

# Import Matplotlib library
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Define the Unit Step Function
# ------------------------------------------------------------
# u[n] = 1 for n >= 0
# u[n] = 0 for n < 0
def u(n):
    return np.where(n >= 0, 1, 0)

# ------------------------------------------------------------
# Generate Sample Indices
# ------------------------------------------------------------
# Sample values from -2 to 10
n = np.arange(-2, 11)

# ------------------------------------------------------------
# Generate the Required Signal
# ------------------------------------------------------------
# x[n] = u[n] - u[n-3] - 5u[n-7]
x = u(n) - u(n - 3) - 5 * u(n - 7)

# ------------------------------------------------------------
# Create Figure
# ------------------------------------------------------------
plt.figure(figsize=(9, 5))

# ------------------------------------------------------------
# Plot the Discrete-Time Signal using Stem Plot
# ------------------------------------------------------------
plt.stem(
    n,
    x,
    linefmt='b-',
    markerfmt='bo',
    basefmt='k-'
)

# ------------------------------------------------------------
# Add Graph Title and Axis Labels
# ------------------------------------------------------------
plt.title("Signal 4 : x[n] = u[n] - u[n-3] - 5u[n-7]", fontsize=14)
plt.xlabel("Sample Index (n)", fontsize=12)
plt.ylabel("Amplitude x[n]", fontsize=12)

# ------------------------------------------------------------
# Display Student Name
# ------------------------------------------------------------
plt.text(4.5, 1.2, "Shivkumar Paun", fontsize=10)

# ------------------------------------------------------------
# Set Axis Limits
# ------------------------------------------------------------
plt.xlim(-2.5, 10.5)
plt.ylim(-6, 2)

# ------------------------------------------------------------
# Display Integer Tick Marks
# ------------------------------------------------------------
plt.xticks(np.arange(-2, 11, 1))
plt.yticks(np.arange(-5, 2, 1))

# ------------------------------------------------------------
# Enable Grid
# ------------------------------------------------------------
plt.grid(True, linestyle='--', alpha=0.7)

# ------------------------------------------------------------
# Display the Graph
# ------------------------------------------------------------
plt.show()
```

```python
# Import NumPy library
import numpy as np

# Import Matplotlib library
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Generate Sample Indices
# ------------------------------------------------------------
# Sample points from -3 to 3
n = np.arange(-3, 4)

# ------------------------------------------------------------
# Initialize Signal with Zeros
# ------------------------------------------------------------
# Create an array of zeros
x = np.zeros(len(n))

# ------------------------------------------------------------
# Define Unit Impulse Signal
# ------------------------------------------------------------
# x(n) = δ(n) + 3δ(n-1) + 5δ(n+1)
x[n == -1] = 5
x[n == 0] = 1
x[n == 1] = 3

# ------------------------------------------------------------
# Create Figure
# ------------------------------------------------------------
plt.figure(figsize=(8, 5))

# ------------------------------------------------------------
# Plot the Signal using Stem Plot
# ------------------------------------------------------------
plt.stem(n, x)

# ------------------------------------------------------------
# Add Graph Title and Labels
# ------------------------------------------------------------
plt.title("Signal 5 : Unit Impulse Signal")
plt.xlabel("Sample Index (n)")
plt.ylabel("Amplitude x[n]")

# ------------------------------------------------------------
# Display Student Name
# ------------------------------------------------------------
plt.text(-2.5, 5.3, "Shivkumar Paun", fontsize=10)

# ------------------------------------------------------------
# Enable Grid
# ------------------------------------------------------------
plt.grid(True)

# ------------------------------------------------------------
# Display Graph
# ------------------------------------------------------------
plt.show()
```
