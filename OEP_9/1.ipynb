```python
import cv2
import requests
import matplotlib.pyplot as plt

# Download image directly from URL
url = "https://images.unsplash.com/photo-1449824913935-59a10b8d2000"

data = requests.get(url).content

with open("input.jpg", "wb") as f:
    f.write(data)

# Read image
image = cv2.imread("input.jpg")

# Smoothing
average = cv2.blur(image, (5, 5))
gaussian = cv2.GaussianBlur(image, (5, 5), 0)
median = cv2.medianBlur(image, 5)
bilateral = cv2.bilateralFilter(image, 9, 75, 75)

# Laplacian sharpening
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
laplacian = cv2.Laplacian(gray, cv2.CV_64F)
laplacian = cv2.convertScaleAbs(laplacian)

sharp = cv2.addWeighted(
    image, 1.0,
    cv2.cvtColor(laplacian, cv2.COLOR_GRAY2BGR),
    -0.5, 0
)

# Unsharp masking
blur = cv2.GaussianBlur(image, (5, 5), 0)
unsharp = cv2.addWeighted(image, 1.5, blur, -0.5, 0)

# Display
images = [
    image, average, gaussian, median,
    bilateral, sharp, unsharp
]

titles = [
    "Original",
    "Average",
    "Gaussian",
    "Median",
    "Bilateral",
    "Laplacian Sharpening",
    "Unsharp Masking"
]

plt.figure(figsize=(14, 10))

for i in range(len(images)):
    plt.subplot(3, 3, i + 1)
    plt.imshow(cv2.cvtColor(images[i], cv2.COLOR_BGR2RGB))
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Smoothing filters

average = cv2.blur(image, (5, 5))

gaussian = cv2.GaussianBlur(
    image, (5, 5), 0
)

median = cv2.medianBlur(
    image, 5
)

bilateral = cv2.bilateralFilter(
    image, 9, 75, 75
)

# Display smoothing results

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(average, cv2.COLOR_BGR2RGB))
plt.title("Average Filter")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(cv2.cvtColor(gaussian, cv2.COLOR_BGR2RGB))
plt.title("Gaussian Filter")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(cv2.cvtColor(median, cv2.COLOR_BGR2RGB))
plt.title("Median Filter")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(cv2.cvtColor(bilateral, cv2.COLOR_BGR2RGB))
plt.title("Bilateral Filter")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Laplacian sharpening

gray = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2GRAY
)

laplacian = cv2.Laplacian(
    gray,
    cv2.CV_64F
)

laplacian = cv2.convertScaleAbs(
    laplacian
)

sharp = cv2.addWeighted(
    image,
    1.0,
    cv2.cvtColor(
        laplacian,
        cv2.COLOR_GRAY2BGR
    ),
    -0.5,
    0
)

# Unsharp Masking

blur = cv2.GaussianBlur(
    image,
    (5, 5),
    0
)

unsharp = cv2.addWeighted(
    image,
    1.5,
    blur,
    -0.5,
    0
)

# Display sharpening results

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(sharp, cv2.COLOR_BGR2RGB))
plt.title("Laplacian Sharpening")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(unsharp, cv2.COLOR_BGR2RGB))
plt.title("Unsharp Masking")
plt.axis("off")

plt.tight_layout()
plt.show()
```
